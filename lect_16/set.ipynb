```python
# Q1: WAP to create two sets , A and B and find the elements that are common to both sets.
a = {1,2,3,4,5}
b = {4,5,6,7,8}
print("intersection",a&b)
```

```python
#Q2: WAP to create two sets , X and Y and find the elements that are unique to each set not present in both.
x = {10,20,30,40}
y = {30,40,50,60}
print("symmetric difference",x^y)
```

```python
# Q3: WAP to check if all elements of one set are contained within another set.
set1 = {1,2,3}
set2 = {1,2,3,4,5}
print("Is set1 a subset of set2?", set1.issubset(set2))
```

```python
# Q4: WAP which create a new set from given two sets have only that elements which are not common.
a = {2,4,6}
b = {4,6,8}
print("symmetric difference",a^b)
```

```python
# Q5: WAP to combine all elements from three different sets into one.
set1 = {1,2}
set2 = {2,3}
set3 = {3,4}
combined_set = set1 | set2 | set3
print("Combined set:", combined_set)
```

```python
#Q6: WAP to find all elements that are in one set but not in another.
a = {10,20,30}
b = {20,30,40}
print("symmetric difference",a^b)
```

```python
# Q7: WAP to remove a specific element(n) from a set,but only if it exits in the set.S={1,2,3,4,5} remove 2 and discard 7
S = {1,2,3,4,5}
S.remove(2) 
S.discard(7)
print(S)

```
