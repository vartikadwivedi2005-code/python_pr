In Python, a set is an unordered collection of unique elements. Think of it like a bag of items where duplicates aren't allowed, and the order doesn't matter.

Sets are highly optimized for membership testing (checking if an item exists in the collection) and performing mathematical operations like unions and intersections.

1. Key Characteristics
# Unordered: Items do not have a defined order; you cannot access them by index (e.g., my_set[0] will raise an error).

# Unique: If you try to add a duplicate, Python simply ignores it.

# Unchangeable (Elements): While you can add or remove items from the set, the items themselves must be of an immutable type (like strings, ints, or tuples).

# Mutable (The Set itself): You can grow or shrink the set as needed.

```python
st = {1,2,3,4,5,"hello",6,7,8}  #Set with mixed data types
print(st)
```

```python
# Creation
fruits = {"apple", "banana", "cherry"}

# Adding items
fruits.add("orange")

# Removing items
fruits.remove("banana")  # Raises error if not present
fruits.discard("mango")  # Does NOT raise error if missing
print(fruits)
```

```python
st = {1,2,3,4,5,"hello",6,7,8.1,2,3}  #Set with duplicates
print(st)  # Output will not have duplicates
```

```python
# remove - Raises error if item not found
# discard - Does NOT raise error if item not found

st = {1,2,3,4,5,"hello",6,7,8,(1,2,3)} 
st.remove(3)    # Removes 3 from the set
print(st)
```

```python
st = {1,2,3,4,5,"hello",6,7,8,(1,2,3)} 
st.discard(30)
print(st)
```

```python
del st  # Deletes the entire set
print(st)   # Raises NameError since st is deleted
```

```python
st = {3,4,5,6,7,"hello",6,7,8,(3,4,5)}
st.add(10)   #Adds 10 to the set
st.add((3,60,5))
print(st)
```

```python
st = {3,4,5,6,7,"hello",6,7,8,(3,4,5)}
st.update((20,30,9))   # Adds multiple items to the set
print(st)
```

```python
st = {1,3,4,5}
print(st)
st.add((3,4,5))    #output - {1, 3, 4, 5, (3, 4, 5)}
print(st)
# st.add([20,30,50])
# print(st)
st.update((3,4,500))   # output - {1, 3, 4, 5, (3, 4, 5), (20, 30, 50), 500}
print(st)
st.update([20,30,400])    # output - {1, 3, 4, 5, (3, 4, 5), (20, 30, 50), 400, 500}
print(st)
st.add([20,30,50])   # Raises TypeError since lists are unhashable and cannot be added to a set
print(st)
```

```python
st = {2,3,4,5,5,6,7,8}
st1 = {3,4,5,6,7,8,9}
print("union",st|st1)     # Items in either st or st1 or both
```

```python
st = {2,3,4,5,5,6,7,8}
st1 = {3,4,5,6,7,8,9}
print("intersection",st&st1)    # Items common to both sets
```

```python
st = {2,3,4,5,5,6,7,8}
st1 = {3,4,5,6,7,8,9}
print("difference",st-st1)  # Items in st but not in st1
```

```python
st = {2,3,4,5,5,6,7,8}
st1 = {3,4,5,6,7,8,9}
print("symmetric difference",st^st1)  # Items in st or st1 but not in both
```
