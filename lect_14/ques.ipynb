```python
# Q3. Write a python program that takes a string and a character as input and returns a list of indices where the character occurs in the string.
x = "programming"
y = "m"

# Output : [6, 7]
def find_indices(string, char):
    indices = []
    for i in range(len(string)):
        if string[i] == char:
            indices.append(i)
    return indices

find_indices(x, y)
```

```python
# WAP to find all unique characters in a given string
# Output : ['p', 'o', 'r', 'g', 'a', 'm', 'i', 'n']
x = "programming"
def unique_characters(string):
    unique_chars = set()
    for char in string:
        unique_chars.add(char)
    return list(unique_chars)
unique_characters(x)
result = unique_characters(x)
print(result)





```

```python
# Wap to count the number of words in a given string that start with a specific letter which was taken as a input.
x = "This is a string for testing"
y = "t"
def count_words_starting_with_letter(string, letter):
    words = string.split()
    count = 0
    for word in words:
        if word.lower().startswith(letter.lower()):
            count += 1
    return count
count_words_starting_with_letter(x, y)




```

```python
# WAP to remove given special characters from a given string.
x = "Hello@World!123"
# output : "HelloWorld123"
special_characters = "@!"
result = x
for char in special_characters:
    result = result.replace(char, "")
print(result)

```
